INTRODUCCIÓN

En la siguiente red neuronal se muestra la predicción con el operador logico NOT, la compuerta NOT opera sobre una sola entrada y produce como salida el valor opuesto o complementario.

DESARROLLO

```python
# 1. Datos NOT (1 sola entrada y 1 salida)
import numpy as np

X = np.array([[0], [1]])
y = np.array([[1], [0]])
```

```python
# 2. Definición del modelo (input_dim=1 porque solo recibimos un valor por muestra)
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

modelo = Sequential([
    Dense(4, activation='relu', input_dim=1),
    Dense(1, activation='sigmoid')
])
```

```python
# 3. Configuración del optimizador con tasa de aprendizaje
from tensorflow.keras.optimizers import Adam

learning_rate = 0.01
adam_optimizer = Adam(learning_rate=learning_rate)
```

```python
# 4. Compilación del modelo
modelo.compile(
    optimizer=adam_optimizer,
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
# 5. Entrenamiento (aumentamos un poco las épocas ya que la red es pequeña)
history = modelo.fit(
    X, y,
    epochs=100,
    verbose=0,
)
```

```python
# 6. Gráfica de la pérdida (Loss)
import matplotlib.pyplot as plt

plt.plot(history.history['loss'])
plt.title('Evolución de la pérdida (Loss) - Compuerta NOT')
plt.xlabel('Época')
plt.ylabel('Binary Cross-Entropy')
plt.show()
```

```python
# 7. Evaluación del modelo
loss, accuracy = modelo.evaluate(X, y)
print(f'Loss: {loss}, Accuracy: {accuracy}')
```

```python
# 8. Predicciones finales
pred = (modelo.predict(X) > 0.5).astype(int)
for i, (inp, p) in enumerate(zip(X, pred)):
    print(f"{inp[0]} -> {p[0]}")
```
